## 04. Electricity Bill
### *Write a program to calculate an electricity bill based on the slab rates.*

```python
units = int(input("Enter electricity units consumed: "))

bill = 0

# 1.100 units → ₹5 per unit
#2.Next 100 units (101–200) → ₹7 per unit
#3.Above 200 units → ₹10 per unit

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = (100 * 5) + (units - 100) * 7
else:
    bill = (100 * 5) + (100 * 7) + (units - 200) * 10

print("Electricity Bill")
print(f"Units Consumed: {units}")
print(f"Total Bill: ₹{bill}")
```

## 05.Armstrong no.
### *Write a program to ckeck whether the number is an Armstrong number.*

```python
# An Armstrong number (also called narcissistic number) is a number equal to the sum of its own digits raised to the power of the number of digits.

#Example:153

num = int(input("Enter a number: "))
sum_of_powers = 0
temp = num
digits = len(str(num))   # Count how many digits

while temp > 0:
    digit = temp % 10
    sum_of_powers += digit ** digits
    temp //= 10

if sum_of_powers == num:
    print(f"{num} is an Armstrong number")
else:
    print(f"{num} is NOT an Armstrong number")
```

## 07.Character classification
### Count the no. of vowel, consonant, digits and special characters in a string 

```python
s = input("Enter a string: ")
vowels = consonants = digits = special = 0

vowel_set = "aeiouAEIOU"

for ch in s:
    if ch.isalpha():  # Check if character is a letter
        if ch in vowel_set:
            vowels += 1
        else:
            consonants += 1
    elif ch.isdigit():  # Check if character is a digit
        digits += 1
    else:  # Anything else is special character
        special += 1
        
print("String Analysis")
print(f"Vowels: {vowels}")
print(f"Consonants: {consonants}")
print(f"Digits: {digits}")
print(f"Special Characters: {special}")
```

## Q9. Longest Word

### Write a function that accepts a sentence and returns the longest word.

```python
def longest_word(sentence):
    words = sentence.split()
    
    longest = words[0]
    
    for word in words:
        if len(word) > len(longest):
            longest = word
    
    return longest
sentence = "I love learning Python programming"

result = longest_word(sentence)

print("Longest word:", result)
```

## Q10. Merge & Remove Duplicates

### Write a function that merges two lists and removes duplicate elements.

```python
def merge_lists(list1, list2):
    merged = list1 + list2
    result = []

    for item in merged:
        if item not in result:
            result.append(item)

    return result
list1 = [121, 672, 893, 454]
list2 = [893, 404, 325, 986]

result = merge_lists(list1, list2)

print("Merged list without duplicates:", result)
```

## Q12. Missing Number


### Find the missing number from a list containing numbers from 1 to N.

```python
def find_missing(numbers, n):
    total = n * (n + 1) // 2
    actual_sum = sum(numbers)

    return total - actual_sum
numbers = [1, 2, 3, 5, 6]
n = 6
result = find_missing(numbers, n)

print("Missing number:", result)
```

## Q1. Factorial Using Recursion

### Find the factorial of a number using recursion.

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)
number = int(input("Enter a number :"))
result = factorial(number)

print("Factorial:", result)
```

## Q2. Fibonacci Using Recursion

### Generate the Fibonacci sequence using recursion.

```python
def fibonacci(n):
    if n <= 1:
        return n
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)
n = int(input("Enter a number :"))
for i in range(n):
    print(fibonacci(i), end=" ")
```

## Q3. Sum of Digits Using Recursion

### Find the sum of digits of a number using recursion.

```python
def sum_digits(n):
    if n == 0:
        return 0
    else:
        return (n % 10) + sum_digits(n // 10)


number = int(input("Enter a number :"))

result = sum_digits(number)

print("Sum of digits:", result)
```

## Q5. Tower of Hanoi

### Solve the Tower of Hanoi problem using recursion.

```python
def tower_of_hanoi(n, source, auxiliary, destination):
    if n == 1:
        print("Move disk 1 from", source, "to", destination)
        return

    tower_of_hanoi(n - 1, source, destination, auxiliary)

    print("Move disk", n, "from", source, "to", destination)

    tower_of_hanoi(n - 1, auxiliary, source, destination)


n = 3

tower_of_hanoi(n, "A", "B", "C")
```

## Q8. Numbers Until Stop

### Read numbers until "stop" is entered, handle invalid inputs using try-except, then calculate minimum, maximum, average, and median.

```python
numbers = []

while True:
    value = input("Enter a number or type 'stop': ")

    if value.lower() == "stop":
        break

    try:
        number = float(value)
        numbers.append(number)

    except ValueError:
        print("Invalid input. Please enter a number.")


if numbers:
    numbers.sort()

    minimum = min(numbers)
    maximum = max(numbers)
    average = sum(numbers) / len(numbers)

    n = len(numbers)

    if n % 2 == 1:
        median = numbers[n // 2]
    else:
        median = (numbers[n // 2 - 1] + numbers[n // 2]) / 2

    print("Minimum:", minimum)
    print("Maximum:", maximum)
    print("Average:", average)
    print("Median:", median)

else:
    print("No numbers were entered.")
```

## Q9. Flatten a Nested List

### Write a recursive function to flatten a nested list.

```python
def flatten_list(nested_list):
    result = []

    for item in nested_list:
        if isinstance(item, list):
            result.extend(flatten_list(item))
        else:
            result.append(item)

    return result


numbers = [12, [72, 23], [74, [85, 76]]]

result = flatten_list(numbers)

print("Flattened list:", result)
```

## Q10. Student Record Management System

### Create a Student Record Management System using function, lists, loops, list comprehensions, exeception handling and file handling.

```python
students = []


# Add a student
def add_student():
    try:
        name = input("Enter student name: ")
        age = int(input("Enter student age: "))
        marks = float(input("Enter student marks: "))

        student = [name, age, marks]
        students.append(student)

        print("Student added successfully.")

    except ValueError:
        print("Please enter valid age and marks.")


# Display all students
def display_students():
    if not students:
        print("No student records found.")
        return

    for student in students:
        print("Name:", student[0])
        print("Age:", student[1])
        print("Marks:", student[2])
        print("-------------------")


# Search for a student
def search_student():
    name = input("Enter student name to search: ")

    result = [student for student in students
              if student[0].lower() == name.lower()]

    if result:
        print("Student found:", result[0])
    else:
        print("Student not found.")


# Save students to a file
def save_students():
    try:
        with open("students.txt", "w") as file:
            for student in students:
                file.write(
                    f"{student[0]},{student[1]},{student[2]}\n"
                )

        print("Records saved successfully.")

    except Exception as e:
        print("Error while saving:", e)


# Main menu
while True:

    print("\n--- Student Record Management System ---")
    print("1. Add Student")
    print("2. Display Students")
    print("3. Search Student")
    print("4. Save Records")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_student()

    elif choice == "2":
        display_students()

    elif choice == "3":
        search_student()

    elif choice == "4":
        save_students()

    elif choice == "5":
        print("Program ended.")
        break

    else:
        print("Invalid choice.")
```
